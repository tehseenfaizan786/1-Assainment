```python

# Python String Methods Assignment

# Step 1: Create 3 variables and take input

name = input("Enter your name: ")
city = input("Enter your city: ")
text = input("Enter any text or number: ")


# Step 2: Apply String Methods on Name

print("\n--- STRING METHODS ON NAME ---")

print("Upper:", name.upper())
print("Lower:", name.lower())
print("Title:", name.title())
print("Capitalize:", name.capitalize())
print("Swapcase:", name.swapcase())
print("Strip:", name.strip())
print("Left Strip:", name.lstrip())
print("Right Strip:", name.rstrip())
print("Replace:", name.replace("a", "o"))
print("Split:", name.split())
print("Join:", "-".join(name.split()))
print("Find:", name.find("a"))
print("Index:", name.index("a") if "a" in name else "Not Found")
print("Count:", name.count("a"))
print("Startswith:", name.startswith("A"))
print("Endswith:", name.endswith("n"))
print("Is Alpha:", name.isalpha())
print("Is Digit:", name.isdigit())
print("Is Alnum:", name.isalnum())
print("Is Space:", name.isspace())
print("Is Lower:", name.islower())
print("Is Upper:", name.isupper())
print("Is Title:", name.istitle())
print("Center:", name.center(20, "*"))
print("Left Justify:", name.ljust(20, "-"))
print("Right Justify:", name.rjust(20, "-"))
print("Zfill:", name.zfill(10))
print("Partition:", name.partition("a"))
print("Rpartition:", name.rpartition("a"))
print("Casefold:", name.casefold())


# Step 3: Apply String Methods on City

print("\n--- STRING METHODS ON CITY ---")

print("Upper:", city.upper())
print("Lower:", city.lower())
print("Title:", city.title())
print("Capitalize:", city.capitalize())
print("Swapcase:", city.swapcase())
print("Strip:", city.strip())
print("Left Strip:", city.lstrip())
print("Right Strip:", city.rstrip())
print("Replace:", city.replace("a", "o"))
print("Split:", city.split())
print("Join:", "-".join(city.split()))
print("Find:", city.find("a"))
print("Index:", city.index("a") if "a" in city else "Not Found")
print("Count:", city.count("a"))
print("Startswith:", city.startswith("K"))
print("Endswith:", city.endswith("n"))
print("Is Alpha:", city.isalpha())
print("Is Digit:", city.isdigit())
print("Is Alnum:", city.isalnum())
print("Is Space:", city.isspace())
print("Is Lower:", city.islower())
print("Is Upper:", city.isupper())
print("Is Title:", city.istitle())
print("Center:", city.center(20, "*"))
print("Left Justify:", city.ljust(20, "-"))
print("Right Justify:", city.rjust(20, "-"))
print("Zfill:", city.zfill(10))
print("Partition:", city.partition("a"))
print("Rpartition:", city.rpartition("a"))
print("Casefold:", city.casefold())


# Step 4: Apply String Methods on Text

print("\n--- STRING METHODS ON TEXT ---")

print("Upper:", text.upper())
print("Lower:", text.lower())
print("Title:", text.title())
print("Capitalize:", text.capitalize())
print("Swapcase:", text.swapcase())
print("Strip:", text.strip())
print("Left Strip:", text.lstrip())
print("Right Strip:", text.rstrip())
print("Replace:", text.replace("1", "2"))
print("Split:", text.split())
print("Join:", "-".join(text.split()))
print("Find:", text.find("1"))
print("Index:", text.index("1") if "1" in text else "Not Found")
print("Count:", text.count("1"))
print("Startswith:", text.startswith("1"))
print("Endswith:", text.endswith("1"))
print("Is Alpha:", text.isalpha())
print("Is Digit:", text.isdigit())
print("Is Alnum:", text.isalnum())
print("Is Space:", text.isspace())
print("Is Lower:", text.islower())
print("Is Upper:", text.isupper())
print("Is Title:", text.istitle())
print("Center:", text.center(20, "*"))
print("Left Justify:", text.ljust(20, "-"))
print("Right Justify:", text.rjust(20, "-"))
print("Zfill:", text.zfill(10))
print("Partition:", text.partition("1"))
print("Rpartition:", text.rpartition("1"))
print("Casefold:", text.casefold())
```
